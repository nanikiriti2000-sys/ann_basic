```python
df = pd.read_csv('train.txt', sep=';', header=None, names=['text', 'emotion'])

# Display the unique string emotions before mapping to confirm
print("Original unique emotions:", df['emotion'].unique())

unique_emotions = df['emotion'].unique()
emotion_numbers = {}
i = 0
for emo in unique_emotions:
  emotion_numbers[emo] = i
  i += 1

# Map string emotions to numerical values
df['emotion'] = df['emotion'].map(emotion_numbers)

print("Emotion mapping:", emotion_numbers)

# Display the first few rows with the new numerical emotions
display(df.head())
```

**lowercase**

```python
df['text']=df['text'].apply(lambda x:x.lower())
```

**remove punctuations**

```python
import string
def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
df['text'].apply(remove_punc)
df['text']=df['text'].apply(remove_punc)
```

**remove numbers**

```python
def remove_numbers(txt):
  new=""
  for i in txt:
    if not i.isdigit():
      new=new+i
  return new
df['text']=df['text'].apply(remove_numbers)
```

**remove emojis & special characters**

```python
def remove_emojis(txt):
  new=""
  for i in txt:
    if i.isascii():
      new=new+i
  return new
df['text']=df['text'].apply(remove_emojis)
```

**IMPORTANT Remove stopwords**

```python
# we have to import libraries
import nltk
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
from nltk.tokenize import word_tokenize
nltk.download('punkt')
```

```python
nltk.download('punkt_tab') # Explicitly downloading the missing resource
```

```python
stop_words=set(stopwords.words('english'))
stop_words
```

```python
len(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
  words = word_tokenize(txt) # Tokenize the text into individual words
  filtered_words = [word for word in words if word.lower() not in stop_words] # Filter out stopwords
  return ' '.join(filtered_words) # Join the filtered words back into a sentence with spaces

df['text'] = df['text'].apply(remove)
display(df.head())
```
